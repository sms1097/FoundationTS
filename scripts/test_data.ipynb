```python
def binary_search_predecessor(sorted_list: list[int], value: int) -> int:
    """
    Return the index of the largest element <= value.
    If no such element exists, return -1.
    """
    low = 0
    high = len(sorted_list) - 1
    best_index = -1

    while low <= high:
        mid = (low + high) // 2
        if sorted_list[mid] <= value:
            best_index = mid
            low = mid + 1
        else:
            high = mid - 1

    return best_index


def bisect_predecessor(sorted_list: list[int], value: int) -> int:
    """
    Return the index of the largest element <= value.
    If no such element exists, return -1.
    """
    return bisect.bisect_right(sorted_list, value) - 1

```

```python
import time

def timed_binary_search_predecessor(sorted_list, value):
    start = time.perf_counter_ns()
    idx = binary_search_predecessor(sorted_list, value)
    end = time.perf_counter_ns()
    return idx, end - start
```

```python
import time
import bisect

def timed_bisect_predecessor(sorted_list, value):
    start = time.perf_counter_ns()
    idx = bisect.bisect_right(sorted_list, value) - 1
    end = time.perf_counter_ns()
    return idx, end - start
```

```python
arr = list(range(1_000_000))
x = 543_210

print("Manual:", timed_binary_search_predecessor(arr, x))
print("Bisect:", timed_bisect_predecessor(arr, x))
```

```python
def benchmark_ns(fn, runs=10_000):
    start = time.perf_counter_ns()
    for _ in range(runs):
        fn()
    end = time.perf_counter_ns()
    return (end - start) // runs
```

```python
manual_ns = benchmark_ns(lambda: binary_search_predecessor(arr, x))
bisect_ns = benchmark_ns(lambda: bisect_predecessor(arr, x))

print(f"Manual avg: {manual_ns} ns")
print(f"Bisect avg: {bisect_ns} ns")
```
